```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
#line plot
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, label="sin(x)", color='blue')
plt.title("Sine Function")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()
plt.show()
```
