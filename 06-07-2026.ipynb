```python
class computer6():
    def __init__(self,cpu,ram,price,company):
        print("i5", "8GB")
        self.cpu =cpu
        self.ram = ram
        self.price = price
        self.company = company
    def config(self):
        print("Config is",self.cpu,self.ram,self.price, self.company)
  
com1 = computer6('i5','8gb',50000,'Lenovo')
com2 = computer6('i7','16gb',75000, 'HPE')
com1.config()
com2.config()
```
