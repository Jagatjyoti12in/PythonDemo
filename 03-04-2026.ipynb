```python
### Nested while - while inside while called nested while
```

```python
i =1 # initialise i =1

while i<=5: # Condition 1<5, 2<5, 3<5, 4<5, 5<=5, 6<=5 exit
    print("data science",i)
    j=1
    while j<=4:
        print('technology')
        j=j+1
    i = i+1 # increment 1+1 =2 ,2,3,4,5,6.
    print()
```

```python
i =1 # initialise i =1

while i<=5: # Condition 1<5, 2<5, 3<5, 4<5, 5<=5, 6<=5 exit
    print("data science",i, end="")
    j=1
    while j<=4:
        print('technology',end="")
        j=j+1
    i = i+1 # increment 1+1 =2 ,2,3,4,5,6.
    print()
```

```python
i =1
while i<=2:
    j=0
    while j<=2:
        print(i*j,end="")
        j+=1
    print()
    i+=1
        
```
