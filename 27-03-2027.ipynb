#2. Array Manupulations



```python
a1 = np.array([1,2,3])
reshaped = np.reshape(a1,(1,3))# reshape to 1X3
print("Reshaped array:", reshaped)
```

```python
a1 = np.array([1, 2, 3])
reshaped = np.reshape(a1, (1, 3))  # reshape to 1x3

print("Reshaped array:", reshaped)
```

```python
#Flattened  Array
f1 = np.array([[1,2],[3,4],[5,6],[7,8]])
falttened = np.ravel(f1)
print("falttened array:", falttened)
```

```python
#Transpose an  Array
e1 = np.array([[1,2],[3,4]])
transposed = np.transpose(e1)
print("transposed array:\n", transposed)
```

```python
#Stack array   Vertically
a2 = np.array([1,2])
b2 = a2 = np.array([3,4])
stacked = np.vstack([a2,b2])
print("Stacked array:\n", stacked)
```

```python
#Stack array   Horizontal
a2 = np.array([1,2])
b2 = a2 = np.array([3,4])
stacked = np.hstack([a2,b2])
print("Stacked array:\n", stacked)
```

3 . Mathematical Functions


```python
# add two array

g = np.array([1,2,3,4])
added = np.add(g,2) #add 2 to each eleent 
print("Add 2 to g:",added)
```

```python
# add Square array

g = np.array([1,2,3,4])
squared = np.power(g,2) #squared 2 to each eleent 
print("squared to g:",squared)
```

```python
# add Square roort array
import numpy as np
g = np.array([1, 2, 3, 4])
sqrt_g = np.sqrt(g)  # square root of each element
print("Square root of g: \n", sqrt_g)
```

```python

a3 = np.array([1,2,3,4])
dot_product = np.dot(a3,g) # dot product work in same dimention of matrix
print("Dot product of a3 X g matrix is ",dot_product)
```

4. Statistical Functions

```python
#Mean
s = np.array([1,2,3,4])
mean = np.mean(s)
print("Mean of s:", mean)
```

```python
#Standard Deviation of an array
s = np.array([1,2,3,4])
std_dev = np.std(s)
print("Standard deviation of s:", std_dev)
```

```python
min = np.min(s)
print("Min  of s:", min)
```

```python
max = np.max(s)
print("Max  of s:", max)
```

```python
#Variance of an array
s = np.array([1,2,3,4])
var = np.var(s)
print("Variance of var:", var)
```

7. Boolean and logical fucntion

```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([2, 2, 2, 2])

result = np.logical_and(a > 1, b == 2)

print("Logical AND:", result)
```

```python
import numpy as np

a = np.array([1, 2, 3, 4])

result = np.logical_or(a < 2, a > 3)

print("Logical OR:", result)
```

```python
import numpy as np

a = np.array([True, False, True])

result = np.logical_not(a)

print("Logical NOT:", result)
```

```python
import numpy as np

a = np.array([True, False, True])
b = np.array([True, True, False])

result = np.logical_xor(a, b)

print("Logical XOR:", result)
```
